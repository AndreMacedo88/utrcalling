```python
utr = utr_ref.extract_feature(five_gtf_path, 'five_prime_utr')
utr
```

```python
utr = utr.sort_values(by=['seqname', 'start'])
annotations = ['gene_id', 'score', 'strand', 'transcript_id', 'gene_name']
ann_cols, ann_idx = utr_ref.prepare_annotations_index(annotations)
utr_bed = pybedtools.BedTool.from_dataframe(utr, columns=ann_cols)
utr_reduced = utr_ref.reduce_regions(utr_bed, annotations_index=ann_idx)
```
